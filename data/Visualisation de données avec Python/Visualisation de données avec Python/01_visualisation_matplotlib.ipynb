## Bar Plot

Un bar plot (ou histogramme) est l'un des types de graphiques les plus courants. Il montre la relation entre une variable numérique et une variable catégorielle. Chaque entité de la variable catégorielle est représentée par une barre. La taille de la barre représente sa valeur numérique.

```python
plt.style.use('ggplot')
```

```python
# Données d'exemple
etiquettes = ['Categorie 1', 'Categorie 2', 'Categorie 3', 'Categorie 4', 'Categorie 5']
valeurs = [25, 32, 18, 22, 11]

# Créer la figure et les axes
fig, ax = plt.subplots(figsize=(10, 6))

# Tracer le graphique à barres
ax.bar(etiquettes, valeurs)

# Ajouter des étiquettes et un titre
ax.set_xlabel('Catégories', fontsize=14)
ax.set_ylabel('Valeurs', fontsize=14)
ax.set_title('Graphique à barres', fontsize=16)

# Rotation des étiquettes des catégories
plt.xticks(rotation=45)

# Afficher le graphique
plt.tight_layout()
plt.show()
```

On peut combiner bar et line plot.

```python
experience = [1,2,3,4,5,6,7,8]

data_scientists_salary = [6500, 9280, 12050, 13200, 16672, 21000, 23965, 29793]

software_engineers_salary = [9020, 12873, 15725, 18000, 19790, 20196, 25769,32000 ]
```

```python
#plt.style.use('tableau-colorblind10')

plt.bar(experience,data_scientists_salary,color="r", label= "Data Scientists")
plt.plot(experience,software_engineers_salary, color="g",label= "Software Engineers" )

plt.title("Salaire des data scientists et des ingénieurs logiciels en fonction de leur expérience")
plt.xlabel("Experience")
plt.ylabel("Salaire")

plt.legend()
plt.grid(False)

plt.show()
```

## Pie Chart

 A **pie chart (ou diagramme circulaire)** est un graphique statistique circulaire, qui est divisé en segments pour illustrer une proportion numérique. Dans un diagramme circulaire, la longueur de l'arc de chaque segment (et par conséquent son angle central et sa surface) est proportionnelle à la quantité qu'il représente.

```python
experience = [1,2,3,4,5,6,7,8]

data_scientists_salary = [6500, 9280, 12050, 13200, 16672, 21000, 23965, 29793]

software_engineers_salary = [9020, 12873, 15725, 18000, 19790, 20196, 25769,32000 ]
```

```python
plt.title("Example de Pie Chart")

slices = [60,40]

plt.pie(slices)

plt.tight_layout()

plt.show()
```

Sum of Values for wedges can be different than 100. The size of each wedge is determined by comparing the value with all the other values, by using this formula:

*The value divided by the sum of all values: x/sum(x)*

```python
list_1 = [40,56,72,38,4]

plt.pie(list_1)

plt.show()
```

On peut ajouter des labels.

```python
incomes = [40,56,72,38,4]

persons = ["Josh","Berkay","Maria","Michael","Anastacia"]

plt.pie(incomes,labels=persons)

plt.show()
```

Par défaut, le tracé du premier secteur commence à partir de l'axe des abscisses et se déplace dans le sens antihoraire :

Mais vous pouvez changer l'angle de départ en spécifiant un paramètre `startangle`. Le paramètre startangle est défini avec un angle en degrés, l'angle par défaut étant 0.

```python
incomes = [40,56,72,38,4]

persons = ["Josh","Berkay","Maria","Michael","Anastacia"]

plt.pie(incomes,labels=persons,startangle=180)

plt.show()
```

Si nous voulons mettre en évidence l'un des secteurs, nous pouvons utiliser le paramètre explode. Il prend un tableau avec une valeur pour chaque secteur.

```python
incomes = [40,56,72,38,4]

persons = ["Josh","Berkay","Maria","Michael","Anastacia"]

myexplode = [0 ,0.1, 0, 0, 0]

plt.pie(incomes, labels=persons, startangle=180, explode=myexplode)

plt.show()
```

Nous pouvons également définir la couleur de chaque secteur avec le paramètre colors.

## Stack Plot

L'idée des graphiques empilés est de montrer les 'parties d'un tout' au fil du temps ; essentiellement, c'est comme un diagramme circulaire, mais dans le temps.

On peut utiliser la fonction **stackplot()**.

```python
days = [1,2,3,4,5,6]

sleep = [6,7,5,8,6,7]
drinking_water = [2,2,1,2,1,1]
work = [5,7,10,8,6,9]
exercise = [3,3,0,1,3,2]
```

```python
plt.style.use("fivethirtyeight")

plt.plot([], [], label='sleep', linewidth=3)
plt.plot([],[], label='drinking_water', linewidth=3)
plt.plot([],[], label='work', linewidth=3)
plt.plot([],[], label='play', linewidth=3)

plt.stackplot(days, sleep, drinking_water, work, exercise)

plt.xlabel('days')
plt.ylabel('activities')
plt.title('6 DAY ROUTINE STACK PLOT EXAMPLE')
plt.legend(loc="lower right")

plt.tight_layout()
plt.show()
```

Nous pouvons visualiser les données qui ont un total spécifique.

```python
stock1= [5,3,3,6,1,8,2,7,9]
stock2= [2,4,1,3,5,0,3,1,0]
stock3= [2,2,5,1,3,1,4,1,0]

days =[1,2,3,4,5,6,7,8,9]
```

```python
stocks= ["stock1","stock2","stock3"]
colors = ["#F9CDAD", "#FC9D9A", "#83AF9B"]

plt.title("Stack Plot of Stock Rates")

plt.stackplot(days, stock1, stock2, stock3, labels=stocks, colors=colors)

plt.legend()
plt.tight_layout()
plt.show()
```

## Histograms

Un histogramme est un graphique montrant des distributions de fréquences.

C'est un graphique montrant le nombre d'observations dans chaque intervalle donné.

Nous utilisons la fonction `hist()` pour créer des histogrammes.

```python
notes = [30,74,94,14,55,47,63,28,88,44,53,18,66,74,81]
```

```python
plt.style.use("fivethirtyeight")

plt.hist(notes, bins=10)

plt.show()
```

Nous pouvons ajouter des couleurs de bordure pour mieux interpréter le graphique.

```python
plt.style.use("fivethirtyeight")

plt.hist(notes,edgecolor="black", bins=5)

plt.title("Notes")
plt.xlabel("Notes")
plt.ylabel("Person")

plt.tight_layout()
plt.grid(False)
plt.show()
```

We can specify the size of bins.

```python
plt.style.use("fivethirtyeight")

plt.hist(notes, bins=5, color="g", edgecolor="black")

plt.title("Notes")
plt.xlabel("Notes")
plt.ylabel("Person")

plt.tight_layout()
plt.grid(False)
plt.show()
```

Traçon une distribution normal

```python
x = np.random.normal(170, 10, 250)

plt.hist(x,color="y",edgecolor="black")

plt.title("Normal Distribution")
plt.xlabel("Numbers")
plt.ylabel("Count")

plt.tight_layout()
plt.grid(False)
plt.show()
```

L'ensemble de données peut être téléchargé à partir d'ici : https://www.kaggle.com/rhuebner/human-resources-data-set

## Scatter Plots

Scatter plots ou les diagrammes de dispersion sont utilisés pour représenter des points de données sur un axe horizontal et vertical dans le but de montrer dans quelle mesure une variable est affectée par une autre.

```python
first_exam_grades = [89, 90, 70, 89, 100, 80, 90, 100, 80, 34]
second_exam_grades = [30, 29, 49, 48, 100, 48, 38, 45, 20, 30]
```

```python
plt.title("Exam Grades Scatter plot")

plt.scatter(first_exam_grades,second_exam_grades)

plt.tight_layout()
plt.xlabel("First Exam Grades")
plt.ylabel("Second Exam Grades")
plt.grid(True)
plt.show()
```

Nous pouvons changer la taille et la couleur des points.

```python
plt.title("Exam Grades Scatter plot")

plt.scatter(first_exam_grades,second_exam_grades,s=100,color="r")

plt.tight_layout()
plt.xlabel("First Exam Grades")
plt.ylabel("Second Exam Grades")
plt.grid(True)
plt.show()
```

Vous pouvez changer la taille des points en fonction des valeurs.

```python
plt.title("Exam Grades Scatter plot")

sizes = np.array([20,50,100,200,500,1000,60,90,10,300])

plt.scatter(first_exam_grades,second_exam_grades,s=sizes,color="r")

plt.tight_layout()
plt.xlabel("First Exam Grades")
plt.ylabel("Second Exam Grades")
plt.grid(True)
plt.show()
```

Nous pouvons également changer le marqueur `marker`.

```python
plt.title("Exam Grades Scatter plot")

plt.scatter(first_exam_grades,second_exam_grades,s=100,color="green",marker="x")

plt.tight_layout()
plt.xlabel("First Exam Grades")
plt.ylabel("Second Exam Grades")
plt.grid(True)
plt.show()
```

## Plotting Time Series

Une série temporelle est une séquence de points de données numériques disposés dans un ordre successif. En investissement, une série temporelle suit l'évolution des points de données choisis, tels que le prix d'un titre, sur une période spécifiée avec des points de données enregistrés à intervalles réguliers.

```python
dates = [
    datetime(2024, 5, 10),
    datetime(2024, 5, 13),
    datetime(2024, 5, 14),
    datetime(2024, 5, 15),
    datetime(2024, 5, 16),
    datetime(2024, 5, 17),
    datetime(2024, 5, 18),
    datetime(2024, 5, 19)
]

values = [0,3,4,7,5,3,5,6]
```

```python
plt.title("Time Series")

plt.plot_date(dates, values)

plt.xticks(rotation='vertical')

plt.show() 
```

Nous pouvons ajouter une ligne

```python
plt.title("Time Series")

plt.plot_date(dates, values, linestyle="solid", marker='o', ms=20, mfc='r', c="b" )

plt.xticks(rotation='vertical')
plt.xlabel("Dates")
plt.ylabel("Values")
plt.grid(False)

plt.show() 
```

## Heatmap

Il est souvent souhaitable de montrer des données qui dépendent de deux variables indépendantes sous forme d'un graphique d'image codé en couleurs. Cela est souvent appelé une heatmap. Si les données sont catégoriques, cela s'appellerait une heatmap catégorique.

Une heatmap est une technique de visualisation de données qui montre la magnitude d'un phénomène en couleur dans deux dimensions. La variation de couleur peut être par teinte ou intensité, donnant des indications visuelles évidentes au lecteur sur la façon dont le phénomène est regroupé ou varie dans l'espace. Elle est généralement utilisée pour comprendre les correlations entre les variables.

Les fonctions `imshow()` ou `heatmap()` de Matplotlib rendent la production de tels graphiques particulièrement facile. matplotlib.pyplot.pcolormesh() est une fonction alternative.

```python
data = np.random.random((6 , 6)) 
data
```

```python
plt.imshow( data , cmap = 'autumn' ) 
  
plt.title( "2-D Heat Map" ) 
plt.show() 
```

```python
sns.heatmap( data , linewidth = 0.5 , cmap = 'coolwarm' ) 
  
plt.title( "2-D Heat Map" ) 
plt.show() 
```

```python
plt.pcolormesh( data , cmap = 'summer' ) 
  
plt.title( '2-D Heat Map' ) 
plt.show() 
```
